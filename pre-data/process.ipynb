```python
from datasets import load_dataset

rels = load_dataset(
    "th1nhng0/vietnamese-legal-documents",
    "relationships",
    split="data"
)
```

```python
from datasets import load_dataset

meta = load_dataset(
    "th1nhng0/vietnamese-legal-documents",
    "metadata",
    split="data"
)
```

```python
meta_df = meta.to_pandas()
rels_df = rels.to_pandas()
```

```python
meta_df.to_parquet(
    "legal_metadata.parquet",
    index=False
)
```

```python
graph = rels_df.merge(
    meta_df,
    left_on="doc_id",
    right_on="id",
    how="left"
)

graph = graph.rename(columns={
    "title": "doc_title",
    "so_ky_hieu": "doc_so_ky_hieu",
    "ngay_ban_hanh": "doc_ngay_ban_hanh",
    "loai_van_ban": "doc_loai_van_ban",
    "ngay_co_hieu_luc": "doc_ngay_co_hieu_luc",
    "ngay_het_hieu_luc": "doc_ngay_het_hieu_luc",
    "nganh": "doc_nganh",
    "linh_vuc": "doc_linh_vuc",
    "co_quan_ban_hanh": "doc_co_quan_ban_hanh",
    "tinh_trang_hieu_luc": "doc_tinh_trang_hieu_luc",
    "year": "doc_year"
})

graph = graph.drop(columns=["id"])
```

```python
graph.head()
```

```python
meta_df["id"] = meta_df["id"].astype(str)

graph["doc_id"] = graph["doc_id"].astype(str)
graph["other_doc_id"] = graph["other_doc_id"].astype(str)
```

```python
graph = graph.merge(
    meta_df,
    left_on="other_doc_id",
    right_on="id",
    how="left",
    suffixes=("", "_other")
)
```
